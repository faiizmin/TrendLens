3 Data Cleaning

```python
#3.1 Check duplicates

print("Duplicates:", df.duplicated().sum())
```

```python
#Remove duplicate if any
df = df.drop_duplicates()
```

```python
import plotly.express as px
```

```python
fig = px.line(
    df,
    x= "Timestamp",
    y= "Temperature_C",
    title = "Temperature Trend"
)

fig.show()
```

```python
day_1 = df[
    df["Timestamp"] < "2026-01-02"
]
```

```python
fig = px.line(
    day_1,
    x="Timestamp",
    y="Temperature_C",
    title="Temperature Trend - Day 1"
)

fig.show()
```

```python
fig.add_hline(
    y=185,
    line_dash="dash",
    annotation_text="Target = 185 C"
)

fig.show()
```

```python
fig.add_hline(
    y=190,
    line_dash="dash",
    annotation_text="Upper limit"
)

fig.add_hline(
    y=180,
    line_dash="dash",
    annotation_text="Lower limit"
)
fig.show()
```

```python
event_period = df[
    (df["Timestamp"] >= "2026-01-05 14:30") &
    (df["Timestamp"] <= "2026-01-05 17:30")
]
```

```python
fig = px.line(
    event_period,
    x="Timestamp",
    y="Temperature_C",
    title="High Temperature Event Investigation"
)

fig.add_hline(
    y=190,
    line_dash="dash",
    annotation_text="Upper limit"
)

fig.show()
```

```python
fig = px.line(
    event_period,
    x="Timestamp",
    y="Production_kg_hr",
    title="Production During High Temperature Event"
)

fig.show()
```

```python
from plotly.subplots import make_subplots
import plotly.graph_objects as go

fig = make_subplots(specs=[[{"secondary_y": True}]])

fig.add_trace(
    go.Scatter(
        x=event_period["Timestamp"],
        y=event_period["Temperature_C"],
        name="Temperature"
    ),
    secondary_y=False
)

fig.add_trace(
    go.Scatter(
        x=event_period["Timestamp"],
        y=event_period["Production_kg_hr"],
        name="Production"
    ),
    secondary_y=True
)

fig.update_layout(
    title="Temperature vs Production"
)

fig.update_yaxes(
    title_text="Temperature (°C)",
    secondary_y=False
)

fig.update_yaxes(
    title_text="Production (kg/hr)",
    secondary_y=True
)

fig.show()
```
